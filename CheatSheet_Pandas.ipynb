```python
# Keywords: loc, label based indexing, select rows and columns by name, label slicing inclusive
# Answer: df.loc[row_labels, col_labels]
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['NY', 'LA', 'SF']
}, index=['a', 'b', 'c'])

# Select single element by row label and column name
print(df.loc['b', 'Name'])

# Select multiple rows and specific columns by label
print(df.loc[['a', 'c'], ['Name', 'City']])

# Slice by row labels (Note: loc label slicing includes the end boundary 'c')
print(df.loc['a':'c', 'Age':'City'])

# Conditional indexing with loc
print(df.loc[df['Age'] > 28, ['Name', 'City']])

```

```python
# Keywords: isnull, isna, find missing values, detect null values, check NaN values
# Answer: df.isnull() or df.isna()
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [np.nan, 2.5, 3.5, np.nan],
    'C': ['cat', 'dog', None, 'bird']
})

# Returns boolean DataFrame of same shape (True where missing)
print(df.isnull())  # Equivalent to df.isna()

# Keywords: count null values, total missing values per column, missing data summary
# Answer: df.isnull().sum()
print(df.isnull().sum())

# Keywords: notnull, notna, find non missing values, filter non null rows
# Answer: df.notnull() or df.notna()
print(df.notnull())

# Keywords: dropna, remove missing values, drop null rows, drop missing columns
# Answer: df.dropna(axis=0, how='any', thresh=None, subset=None, inplace=False)
print(df.dropna())                         # Drop rows where any value is NaN
print(df.dropna(how='all'))                 # Drop rows where all values are NaN
print(df.dropna(subset=['A', 'B']))         # Drop rows where NaN exists in specific columns
print(df.dropna(axis=1))                    # Drop columns containing any NaN

# Keywords: fillna, replace missing values, impute nulls, fill NaN with value or strategy
# Answer: df.fillna(value=None, method=None, axis=None, inplace=False)
print(df.fillna(value=0))                  # Replace all NaNs with constant value 0
print(df['A'].fillna(df['A'].mean()))      # Impute column NaNs with mean
print(df.ffill())                           # Forward fill (propagate last valid value)
print(df.bfill())                           # Backward fill (propagate next valid value)
```

```python
# Keywords: sum, column sum, row sum, numeric sum
# Answer: df.sum(axis=0, numeric_only=False)
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'A', 'B', np.nan],
    'Values': [10, 20, 10, 30, 20, 50]
})

print(df['Values'].sum())           # Sum of a single Series
print(df.sum(numeric_only=True))    # Sum across all numeric columns (axis=0 for columns)
print(df.sum(axis=1, numeric_only=True)) # Sum across rows

# Keywords: count, count non null, count non missing entries
# Answer: df.count(axis=0)
print(df.count())                   # Counts non-NA/null cells for each column
print(df['Category'].count())       # Counts non-NA/null cells in a single Series

# Keywords: value_counts, frequency count, count unique values, categorical distribution
# Answer: df['col'].value_counts(normalize=False, sort=True, ascending=False, bins=None, dropna=True)
print(df['Category'].value_counts())                       # Count occurrence of each unique value
print(df['Category'].value_counts(dropna=False))            # Include NaN counts in results
print(df['Category'].value_counts(normalize=True))         # Return relative frequencies (percentages)
print(df['Values'].value_counts(bins=2))                   # Bin continuous values into discrete groups
```

```python
# Keywords: groupby, group by single column, aggregate group, groupby mean sum min max
# Answer: df.groupby('col').mean()
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Team': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Pos': ['G', 'F', 'G', 'F', 'G', 'F'],
    'Points': [10, 15, 8, 12, 20, 22],
    'Rebounds': [5, 7, 6, 9, 11, 4]
})

# Single column group with simple aggregation
print(df.groupby('Team')['Points'].mean())

# Keywords: groupby multiple columns, multi index group, group by list of columns
# Answer: df.groupby(['col1', 'col2']).sum()
print(df.groupby(['Team', 'Pos']).sum())

# Keywords: groupby agg, apply multiple aggregations, custom column aggregation, agg dictionary
# Answer: df.groupby('col').agg({'col1': 'mean', 'col2': ['sum', 'max']})
print(df.groupby('Team').agg({
    'Points': ['mean', 'max'],
    'Rebounds': 'sum'
}))

# Keywords: groupby without index, as_index False, reset index group
# Answer: df.groupby('col', as_index=False).mean()
print(df.groupby('Team', as_index=False)['Points'].mean())

# Keywords: groupby filter, filter groups by condition, group size filter
# Answer: df.groupby('col').filter(lambda x: condition)
print(df.groupby('Team').filter(lambda x: x['Points'].sum() > 25))

# Keywords: groupby transform, group normalization, broadcast group calculation to original shape
# Answer: df.groupby('col')['col2'].transform('mean')
df['Team_Avg_Points'] = df.groupby('Team')['Points'].transform('mean')
print(df)
```

```python
# Keywords: unique, find unique values, distinct values in series
# Answer: df['col'].unique()
import pandas as pd

df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'C', 'B', 'A'],
    'Values': [10, 20, 10, 30, 20, 40]
})

# Returns array of unique values in order of appearance
print(df['Category'].unique())

# Keywords: nunique, count unique values, number of distinct elements
# Answer: df['col'].nunique(dropna=True)
print(df['Category'].nunique())

# Count unique values across all columns in DataFrame
print(df.nunique())

# Keywords: drop_duplicates, remove duplicate rows, keep first or last duplicate
# Answer: df.drop_duplicates(subset=None, keep='first', inplace=False, ignore_index=False)
print(df.drop_duplicates())                           # Remove full duplicate rows
print(df.drop_duplicates(subset=['Category']))        # Remove duplicates based on specific column
print(df.drop_duplicates(subset=['Category'], keep='last')) # Keep last occurrence instead of first

# Keywords: duplicated, find duplicate rows, boolean mask for duplicates
# Answer: df.duplicated(subset=None, keep='first')
print(df.duplicated(subset=['Category']))            # Returns boolean Series (True for duplicate rows)
```

```python
# Keywords: concat, concatenate dataframes, stack rows vertically, append dataframes
# Answer: pd.concat([df1, df2], axis=0, ignore_index=True)
import pandas as pd

df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})

# Stack vertically (row-wise)
print(pd.concat([df1, df2], axis=0, ignore_index=True))

# Keywords: concat side by side, concatenate columns horizontally
# Answer: pd.concat([df1, df2], axis=1)
df3 = pd.DataFrame({'C': ['C0', 'C1'], 'D': ['D0', 'D1']})
print(pd.concat([df1, df3], axis=1))


# Keywords: merge, database style join, inner left right outer join on column key
# Answer: pd.merge(left, right, how='inner', on='key', left_on=None, right_on=None)
left = pd.DataFrame({'key': ['K0', 'K1', 'K2'], 'A': ['A0', 'A1', 'A2']})
right = pd.DataFrame({'key': ['K0', 'K1', 'K3'], 'B': ['B0', 'B1', 'B3']})

print(pd.merge(left, right, on='key', how='inner'))  # Inner join (default)
print(pd.merge(left, right, on='key', how='left'))   # Left join
print(pd.merge(left, right, on='key', how='right'))  # Right join
print(pd.merge(left, right, on='key', how='outer'))  # Full outer join

# Keywords: merge different column names, merge on multiple keys, handle suffixes
# Answer: pd.merge(df1, df2, left_on='col1', right_on='col2', suffixes=('_x', '_y'))
left_diff = pd.DataFrame({'id_left': ['K0', 'K1'], 'val': [1, 2]})
right_diff = pd.DataFrame({'id_right': ['K0', 'K1'], 'val': [3, 4]})

print(pd.merge(left_diff, right_diff, left_on='id_left', right_on='id_right', suffixes=('_left', '_right')))


# Keywords: join, join on index, index based dataframe join
# Answer: df1.join(df2, how='left', lsuffix='', rsuffix='')
df_a = pd.DataFrame({'A': [1, 2]}, index=['K0', 'K1'])
df_b = pd.DataFrame({'B': [3, 4]}, index=['K0', 'K2'])

print(df_a.join(df_b, how='left'))  # Join based on index
print(df_a.join(df_b, how='outer'))
```

```python
# Keywords: apply, apply lambda to column, column transformation, element wise function
# Answer: df['col'].apply(lambda x: expression)
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': ['apple', 'banana', 'cherry', 'date']
})

# Apply lambda function to transform a single column
df['A_squared'] = df['A'].apply(lambda x: x ** 2)
df['B_length'] = df['B'].apply(lambda x: len(x))
print(df)

# Keywords: apply across rows, row wise lambda, multiple column lambda transformation
# Answer: df.apply(lambda row: func(row['col1'], row['col2']), axis=1)
df['A_times_B_len'] = df.apply(lambda row: row['A'] * len(row['B']), axis=1)
print(df)

# Keywords: map, series map dictionary, element wise replacement, map function to series
# Answer: series.map(dict_or_function)
# Map values using a custom lambda or direct dictionary lookup
df['A_category'] = df['A'].map(lambda x: 'Low' if x <= 2 else 'High')
print(df)

# Keywords: applymap, mapmap, elementwise dataframe function, apply function to whole dataframe
# Answer: df.map(func) or df.applymap(func)  # Note: applymap is deprecated in newer Pandas, use df.map()
numeric_df = pd.DataFrame({'X': [1.234, 5.678], 'Y': [9.101, 2.345]})
print(numeric_df.map(lambda x: f"{x:.1f}"))  # Format all cells across entire DataFrame

# Keywords: custom function pandas, custom function apply, named function transformation
# Answer: df['col'].apply(custom_function_name)
def classify_word(word):
    if len(word) > 5:
        return 'Long'
    return 'Short'

df['Word_Class'] = df['B'].apply(classify_word)
print(df)
```
