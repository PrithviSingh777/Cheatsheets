```python
# Keywords: matplotlib version, check matplotlib version, pyplot version, what version of matplotlib am i using
# Answer: matplotlib.__version__
import matplotlib
import matplotlib.pyplot as plt
print(matplotlib.__version__)
```

```python
# Keywords: bar plot parameters, customize bar chart, bar chart color width edge color hatch align
# Answer: plt.bar(x, height, width, color, edgecolor, linewidth, tick_label, align, hatch, alpha)
import matplotlib.pyplot as plt

x = ['Group A', 'Group B', 'Group C']
heights = [10, 24, 18]

plt.bar(
    x, 
    heights, 
    width=0.6,                 # Width of bars (0.0 to 1.0)
    color=['#3498db', '#e74c3c', '#2ecc71'],  # Fill colors per bar or single color
    edgecolor='black',         # Border color
    linewidth=2,               # Border line thickness
    linestyle='--',            # Border line style (':', '--', '-', '-.')
    hatch='/',                 # Pattern fill ('/', 'o', '*', 'x', '.')
    alpha=0.85,                # Transparency (0.0 transparent to 1.0 opaque)
    align='center'             # Bar alignment relative to x ticks ('center', 'edge')
)
plt.show()
```

```python
# Keywords: scatter plot parameters, customize scatter plot, scatter marker size color cmap alpha edgecolors
# Answer: plt.scatter(x, y, s, c, marker, cmap, norm, vmin, vmax, alpha, linewidths, edgecolors)
#horizontal bars-> just add barh
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 20]
sizes = [50, 100, 150, 200, 250]      # Point sizes in points^2
colors = [10, 20, 30, 40, 50]         # Numerical array for colormap mapping

plt.scatter(
    x, 
    y, 
    s=sizes,                     # Size of markers (single int or array)
    c=colors,                    # Color array (mapped via cmap) or single color string
    marker='o',                  # Marker style ('o', 's', '^', 'D', '*', 'x')
    cmap='viridis',              # Colormap applied to numerical 'c' array
    alpha=0.8,                   # Transparency (0.0 transparent to 1.0 opaque)
    linewidths=1.5,              # Thickness of marker border
    edgecolors='black'           # Border color of markers ('face', 'none', or color string)
)
plt.show()
```

```python
# Keywords: histogram parameters, customize histogram, hist bins range density cumulative histtype color rwidth
# Answer: plt.hist(x, bins, range, density, cumulative, histtype, color, edgecolor, linewidth, alpha, rwidth)
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

plt.hist(
    data, 
    bins=20,                    # Number of equal-width bins or sequence of bin edges
    range=(-3, 3),              # Lower and upper range of the bins (min, max)
    density=True,               # Normalize to display probability density (area integrates to 1)
    cumulative=False,           # Plot cumulative counts/density if True
    histtype='bar',             # Bar type ('bar', 'barstacked', 'step', 'stepfilled')
    align='mid',                # Bin edge alignment ('left', 'mid', 'right')
    orientation='vertical',     # Bar orientation ('vertical', 'horizontal')
    rwidth=0.9,                 # Relative width of bars as fraction of bin width
    color='#2b5c8f',            # Fill color of bars
    edgecolor='black',          # Border color
    linewidth=1.2,              # Border line thickness
    linestyle='-',              # Border line style ('-', '--', ':', '-.')
    alpha=0.75                  # Transparency (0.0 transparent to 1.0 opaque)
)
plt.show()
```

```python
# Keywords: boxplot parameters, customize boxplot, boxplot notch vert patch_artist showmeans meanline boxprops whiskerprops
# Answer: plt.boxplot(x, notch, vert, patch_artist, widths, showmeans, meanline, boxprops, whiskerprops, capprops, flierprops, medianprops)
import matplotlib.pyplot as plt
import numpy as np

data = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.boxplot(
    data, 
    notch=True,                 # Draw notched box plot to indicate confidence intervals of median
    vert=True,                  # Vertical boxes if True, horizontal if False
    patch_artist=True,          # Fill boxes with color (required to use boxprops facecolor)
    widths=0.5,                 # Width of each box
    showmeans=True,             # Display the mean value marker
    meanline=True,              # Draw mean as a line across the box rather than a point
    showfliers=True,            # Display outlier points outside whiskers
    showcaps=True,              # Show caps on ends of whiskers
    showbox=True,               # Show the central box
    boxprops=dict(facecolor='#3498db', color='black', linewidth=1.5),     # Color and style of main box
    whiskerprops=dict(color='black', linewidth=1.5, linestyle='--'),      # Style of whisker lines
    capprops=dict(color='black', linewidth=2),                            # Style of caps on whiskers
    flierprops=dict(marker='o', markerfacecolor='red', markersize=6),     # Style of outlier points
    medianprops=dict(color='yellow', linewidth=2),                         # Style of median line
    meanprops=dict(color='green', linewidth=2)                            # Style of mean line (if meanline=True)
)
plt.show()
```

```python
# Keywords: heatmap parameters, imshow, custom heatmap, imshow aspect origin cmap vmin vmax interpolation
# Answer: plt.imshow(X, cmap, aspect, origin, vmin, vmax, interpolation) with plt.colorbar()
import matplotlib.pyplot as plt
import numpy as np

data = np.random.rand(10, 10)

plt.imshow(
    data, 
    cmap='viridis',             # Colormap ('viridis', 'magma', 'coolwarm', 'plasma')
    aspect='equal',             # Aspect ratio ('equal', 'auto', or float)
    origin='upper',             # Place (0,0) index ('upper' left or 'lower' left)
    vmin=0.0,                   # Minimum data value bound for colormap
    vmax=1.0,                   # Maximum data value bound for colormap
    interpolation='nearest',    # Pixel blending ('nearest', 'bilinear', 'bicubic')
    alpha=1.0                   # Transparency level
)
plt.colorbar(label='Scale')      # Add color scale bar
plt.show()
```

```python
# Keywords: subplots, multiple plots, subplots layout, fig ax subplots, grid of plots
# Answer: plt.subplots(nrows, ncols, figsize, sharex, sharey) / ax.plot()
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# Create a 2x2 grid of subplots
fig, axes = plt.subplots(
    nrows=2, 
    ncols=2, 
    figsize=(10, 8),          # Overall figure width and height in inches
    sharex=False,             # Share X axis limits across subplots (True/False)
    sharey=False,             # Share Y axis limits across subplots (True/False)
    squeeze=True              # Squeeze 2D array of axes into 1D if single row/col
)

# Access individual axes via 2D index axes[row, col]
axes[0, 0].plot(x, np.sin(x), color='blue')
axes[0, 0].set_title('Sine Wave')

axes[0, 1].scatter(x, np.cos(x), color='red', s=10)
axes[0, 1].set_title('Cosine Scatter')

axes[1, 0].bar(['A', 'B', 'C'], [3, 7, 5], color='green')
axes[1, 0].set_title('Bar Chart')

axes[1, 1].hist(np.random.randn(100), bins=15, color='purple')
axes[1, 1].set_title('Histogram')

# Automatically adjust padding between subplots to avoid overlapping text
plt.tight_layout()
plt.show()
```
